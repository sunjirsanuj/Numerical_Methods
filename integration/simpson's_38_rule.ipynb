```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**2

a = 0
b = 2
n = 12

if n % 3 != 0:
    raise ValueError("n must be divisible by 3")

h = (b-a)/n

s = f(a) + f(b)
```

```python
for i in range(1, n):
    if i % 3 == 0:
        s += 2*f(a+i*h)
    else:
        s += 3*f(a+i*h)

result = (3*h/8)*s

exact = 8/3
error = abs(exact-result)

df = pd.DataFrame({
    "Method": ["Simpson's 3/8"],
    "Numerical Result": [result],
    "Exact Result": [exact],
    "Absolute Error": [error]
})

display(df.round(8))
```

```python
x = np.linspace(a, b, 200)

plt.plot(x, f(x), label="f(x) = x²")
plt.fill_between(x, f(x), alpha=0.2)
plt.xlabel("X")
plt.ylabel("f(X)")
plt.title("Simpson's 3/8 Rule")
plt.legend()
plt.grid()
plt.show()

plt.bar(["Numerical", "Exact"], [result, exact])
plt.ylabel("Integral Value")
plt.title("Result Comparison")
plt.grid(axis="y")
plt.show()

plt.bar(["Simpson 3/8"], [error])
plt.ylabel("Absolute Error")
plt.title("Simpson 3/8 Error")
plt.show()
```
