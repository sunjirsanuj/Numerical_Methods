```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**2

a = 0
b = 2
n = 4

h = (b - a) / n

x = np.linspace(a, b, n + 1)
y = f(x)

result = (h / 2) * (
    y[0] + 2 * np.sum(y[1:-1]) + y[-1]
)

exact = 8 / 3
error = abs(exact - result)

df = pd.DataFrame({
    "X": x,
    "Function Value": y
})

display(df.round(8))

summary = pd.DataFrame({
    "Method": ["Trapezoidal Rule"],
    "Numerical Result": [result],
    "Exact Result": [exact],
    "Absolute Error": [error]
})

display(summary.round(8))
```

```python
x = np.linspace(a, b, 200)

plt.plot(x, f(x), label="f(x) = x²")
plt.fill_between(x, f(x), alpha=0.2)
plt.xlabel("X")
plt.ylabel("f(X)")
plt.title("Trapezoidal Method")
plt.legend()
plt.grid()
plt.show()

plt.bar(["Numerical", "Exact"], [result, exact])
plt.ylabel("Integral Value")
plt.title("Result Comparison")
plt.grid(axis="y")
plt.show()

plt.bar(["Trapezoidal"], [error])
plt.ylabel("Absolute Error")
plt.title("Trapezoidal Error")
plt.show()
```
