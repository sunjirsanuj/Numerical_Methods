```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**2

a = 0
b = 2
n = 12

if n % 2 != 0:
    raise ValueError("n must be even")

h = (b-a)/n
```

```python
x = np.linspace(a, b, n + 1)
y = f(x)

weights = np.ones(n + 1)
weights[1:-1:2] = 4
weights[2:-1:2] = 2

result = (h / 3) * np.sum(weights * y)

exact = 8/3
error = abs(exact-result)

df = pd.DataFrame({
    "Method": ["Simpson's 1/3"],
    "Numerical Result": [result],
    "Exact Result": [exact],
    "Absolute Error": [error]
})

display(df.round(8))
```

```python
x = np.linspace(a, b, 200)

plt.plot(x, f(x), label="f(x) = x²")
plt.fill_between(x, f(x), alpha=0.2)
plt.xlabel("X")
plt.ylabel("f(X)")
plt.title("Simpson's 1/3 Rule")
plt.legend()
plt.grid()
plt.show()

plt.bar(["Numerical", "Exact"], [result, exact])
plt.ylabel("Integral Value")
plt.title("Result Comparison")
plt.grid(axis="y")
plt.show()

plt.bar(["Simpson 1/3"], [error])
plt.ylabel("Absolute Error")
plt.title("Simpson 1/3 Error")
plt.show()
```
