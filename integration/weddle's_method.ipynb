```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**2

a = 0
b = 2
n = 12

if n % 6 != 0:
    raise ValueError("n must be divisible by 6")

h = (b-a)/n

total = 0
```

```python
for j in range(0, n, 6):
    total += (
        f(a+j*h)
        + 5*f(a+(j+1)*h)
        + f(a+(j+2)*h)
        + 6*f(a+(j+3)*h)
        + f(a+(j+4)*h)
        + 5*f(a+(j+5)*h)
        + f(a+(j+6)*h)
    )

result = (3*h/10)*total

exact = 8/3
error = abs(exact-result)

df = pd.DataFrame({
    "Method": ["Weddle"],
    "Numerical Result": [result],
    "Exact Result": [exact],
    "Absolute Error": [error]
})

display(df.round(8))
```

```python
x = np.linspace(a, b, 200)

plt.plot(x, f(x), label="f(x) = x²")
plt.fill_between(x, f(x), alpha=0.2)
plt.xlabel("X")
plt.ylabel("f(X)")
plt.title("Weddle's Method")
plt.legend()
plt.grid()
plt.show()

plt.bar(["Numerical", "Exact"], [result, exact])
plt.ylabel("Integral Value")
plt.title("Result Comparison")
plt.grid(axis="y")
plt.show()

plt.bar(["Weddle"], [error])
plt.ylabel("Absolute Error")
plt.title("Weddle Error")
plt.show()
```
