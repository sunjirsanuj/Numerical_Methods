```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def g(x):
    return np.cbrt(x + 2)

x0 = 1.5

tolerance = 0.000001
max_iterations = 100
```

```python
iterations = []
roots = []
absolute_errors = []
relative_errors = []

x = x0

for i in range(1, max_iterations + 1):

    x_new = g(x)

    abs_error = abs(x_new - x)
    rel_error = abs_error / abs(x_new)

    iterations.append(i)
    roots.append(x_new)
    absolute_errors.append(abs_error)
    relative_errors.append(rel_error)

    if abs_error < tolerance:
        x = x_new
        break

    x = x_new
```

```python
result = pd.DataFrame({
    "Iteration": iterations,
    "x": roots,
    "Absolute Error": absolute_errors,
    "Relative Error": relative_errors
})

print(result.to_string(index=False))

print("\nApproximate Root =", roots[-1])
print("Number of Iterations =", len(iterations))
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    iterations,
    absolute_errors,
    marker='o'
)

plt.yscale('log')

plt.xlabel("Iteration")
plt.ylabel("Absolute Error")
plt.title("Fixed-Point Iteration - Error vs Iteration")

plt.grid(True)

plt.show()
```

```python
x_values = np.linspace(0, 2.5, 500)

y_g = g(x_values)
y_x = x_values

root = roots[-1]

plt.figure(figsize=(8, 5))

plt.plot(
    x_values,
    y_g,
    label="g(x) = ∛(x + 2)"
)

plt.plot(
    x_values,
    y_x,
    linestyle="--",
    label="y = x"
)

plt.scatter(
    root,
    root,
    s=80,
    label=f"Fixed Point ≈ {root:.6f}"
)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Fixed-Point Iteration")

plt.legend()
plt.grid(True)

plt.show()
```
