```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def f(x, y):
    return x + y

def exact(x):
    return 2*np.exp(x) - x - 1

x0 = 0
y0 = 1
h = 0.1
n = 10

x = [x0]
y = [y0]
```

```python
for i in range(n):
    xi = x[-1]
    yi = y[-1]

    k1 = f(xi, yi)
    k2 = f(xi+h/2, yi+(h/2)*k1)

    yn = yi + h*k2

    x.append(xi+h)
    y.append(yn)

x = np.array(x)
y = np.array(y)

actual = exact(x)
error = np.abs(actual-y)

df = pd.DataFrame({
    "X": x,
    "Numerical": y,
    "Exact": actual,
    "Absolute Error": error
})

print("Midpoint Method")
display(df.round(8))

print("Final Error:", error[-1])
```

```python
plt.plot(x, actual, label="Exact")
plt.plot(x, y, 'o--', label="Midpoint")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Midpoint Method")
plt.legend()
plt.grid()
plt.show()

plt.plot(x, error, 'r-o')
plt.xlabel("X")
plt.ylabel("Absolute Error")
plt.title("Midpoint Error")
plt.grid()
plt.show()
```
