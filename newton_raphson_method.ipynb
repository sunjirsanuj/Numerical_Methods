```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**3 - x - 2

def df(x):
    return 3*x**2 - 1

x0 = 1.5

tolerance = 0.000001
max_iterations = 100
```

```python
iterations = []
roots = []
function_values = []
absolute_errors = []
relative_errors = []

x = x0

for i in range(1, max_iterations + 1):

    fx = f(x)
    dfx = df(x)

    # Newton-Raphson formula
    x_new = x - fx / dfx

    # Error calculation
    abs_error = abs(x_new - x)
    rel_error = abs_error / abs(x_new)

    iterations.append(i)
    roots.append(x_new)
    function_values.append(f(x_new))
    absolute_errors.append(abs_error)
    relative_errors.append(rel_error)

    # Check convergence
    if abs_error < tolerance:
        x = x_new
        break

    x = x_new
```

```python
result = pd.DataFrame({
    "Iteration": iterations,
    "x": roots,
    "f(x)": function_values,
    "Absolute Error": absolute_errors,
    "Relative Error": relative_errors
})

print(result.to_string(index=False))

print("\nApproximate Root =", roots[-1])
print("Number of Iterations =", len(iterations))
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    iterations,
    absolute_errors,
    marker='o'
)

plt.yscale('log')

plt.xlabel("Iteration")
plt.ylabel("Absolute Error")
plt.title("Newton-Raphson Method - Error vs Iteration")

plt.grid(True)

plt.show()
```

```python
x_values = np.linspace(0, 2.5, 500)
y_values = f(x_values)

root = roots[-1]

plt.figure(figsize=(8, 5))

plt.plot(
    x_values,
    y_values,
    label="f(x) = x³ - x - 2"
)

plt.axhline(
    0,
    linestyle="--"
)

plt.scatter(
    root,
    f(root),
    s=80,
    label=f"Root ≈ {root:.6f}"
)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Newton-Raphson Method - Function and Root")

plt.legend()
plt.grid(True)

plt.show()
```
